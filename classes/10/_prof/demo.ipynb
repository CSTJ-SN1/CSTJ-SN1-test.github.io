```python
i = 0
mots = []

with open("fables.txt") as f:
    texte = f.read()
    mot = ""
    
    while i < len(texte):
        lettre = texte[i]

        if lettre == " " or lettre == "\n":
            if mot != "":
                mots.append(mot)
                mot = ""
        elif lettre == ".":
            if mot != "":
                mots.append(mot)
                mot = ""
            mots.append(".")
        else:
            mot += lettre
            
        i += 1
```

```python
frequences = {}

for i in range(len(mots) - 1):
    mot = mots[i]
    prochain = mots[i + 1]

    if not (mot in frequences):
        frequences[mot] = []

    frequences[mot].append(prochain)
```

```python
import random

phrase = ""
mot = "."

for i in range(100):
    mot = random.choice(frequences[mot])
    phrase += mot + " "

print(phrase)
```
