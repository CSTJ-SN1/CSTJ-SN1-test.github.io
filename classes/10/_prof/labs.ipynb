```python
fruits = {"pomme": 5, "banane": 2, "fraise": 12}

for fruit in fruits:
    if fruits[fruit] > 1:
        print(f"Il reste {fruits[fruit]} {fruit}s.")
    else:
        print(f"Il reste {fruits[fruit]} {fruit}.")
```

```python
texte = "Attack the dog then pretend like nothing happened cats are a queer kind of folk my water bowl is clean and freshly replenished, so i'll drink from the toilet or to pet a cat, rub its belly, endure blood and agony, quietly weep, keep rubbing belly purr as loud as possible, be the most annoying cat that you can, and, knock everything off the table or fall over dead (not really but gets sypathy), flex claws on the human's belly and purr like a lawnmower."
lettres = {}

for lettre in texte:
    if lettre in lettres:
        lettres[lettre] += 1
    else:
        lettres[lettre] = 1

for lettre in lettres:
    print(f"{lettre}: {lettres[lettre]}")
```

```python
noms = {"Paul": 60, "Marie": 22, "Jean": 41, "Sophie": 19, "Arnold": 75}

nouveau = {}

for nom in noms:
    if nom[0] != "A" and nom[0] != "M":
        nouveau[nom] = noms[nom]

print(nouveau)
```
