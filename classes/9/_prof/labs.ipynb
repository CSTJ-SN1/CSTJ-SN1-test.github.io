```python
import random

def lancerDe(faces):
    return random.randint(1, faces)

print(lancerDe(6))
print(lancerDe(8))
print(lancerDe(12))
print(lancerDe(20))
```

```python
def nomComplet(prenom, nom):
    return prenom + " " + nom

print(nomComplet("Eric", "Duranleau"))
```

```python
def printList(coll):
    for item in coll:
        print(item)

noms = ["Frieren", "Fern", "Stark"]

printList(noms)
```

```python
def aireRect(largeur, hauteur):
    return largeur * hauteur

print(aireRect(10, 20))
```

```python
import random

def listeAleatoire(qty, nmin, nmax):
    liste = []
    
    for i in range(qty):
        liste.append(random.randint(nmin, nmax))

    return liste

print(listeAleatoire(5, 0, 10))
```

```python
def moyenne(liste):
    total = 0

    for n in liste:
        total += n

    return total / len(liste)

nums = [10, 20, 25]
print(moyenne(nums))
```

```python
def mediane(liste):
    return liste[int(len(liste) / 2)]

nums = [10, 50, 100, 100, 100]
print(mediane(nums))
```
