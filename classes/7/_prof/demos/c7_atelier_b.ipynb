Jouer avec le dt pour montrer que ce n'est pas tout à fait cyclique

```python
taux_lapin_naissance = 0.3
taux_lapin_mort  = 0.01
taux_renard_naissance = 0.001
taux_renard_mort = 0.05

lapins = 50
renards = 15

historique_lapins = [lapins]
historique_renards = [renards]
historique_etapes = [0]

dt = 0.001
iterations = 100000

for t in range(iterations):
    lapins_nees = taux_lapin_naissance * lapins * dt # naissance naturelle
    lapins_morts = taux_lapin_mort * lapins * renards * dt # mangé par un renard

    renards_nees = taux_renard_naissance * renards * lapins * dt # mangé un lapin
    renards_morts = taux_renard_mort * renards * dt # mort natuelle

    # Update the populations
    lapins += lapins_nees - lapins_morts
    renards += renards_nees - renards_morts

    if lapins < 0:
        lapins = 0

    if renards < 0:
        renards = 0

    historique_lapins.append(lapins)
    historique_renards.append(renards)
    historique_etapes.append(t + 1)

print("Simulation finished.")
```

```python
import matplotlib.pyplot as plt

fig = plt.figure()
ax = fig.add_subplot()

ax.plot(historique_etapes, historique_lapins, label='Lapins', color='blue')
ax.plot(historique_etapes, historique_renards, label='Renards', color='red')

ax.legend()
ax.grid(True)

ax.set_title("Écosystème")
# plt.xlabel('Time')
# plt.ylabel('Population')

plt.show()
```
