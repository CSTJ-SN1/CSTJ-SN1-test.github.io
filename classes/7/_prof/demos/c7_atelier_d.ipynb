```python
import matplotlib.pyplot as plt
import random

steps = 5000
start_x = 0
start_y = 0

x_path = [start_x]
y_path = [start_y]

curr_x = start_x
curr_y = start_y

for i in range(steps):
    
    # ---------------------------------------------------------
    # -- À modifier
    
    dx = random.choice([-1, 1])
    dy = random.choice([-1, 1])
    
    # ---------------------------------------------------------
    
    curr_x += dx
    curr_y += dy
    
    x_path.append(curr_x)
    y_path.append(curr_y)


# ==================================
plt.figure(figsize=(8, 8))

time_values = range(len(x_path))

# 'plasma', 'viridis', 'cool', 'inferno', 'jet'
plt.scatter(x_path, y_path, c=time_values, cmap='plasma', s=5, alpha=0.5)

plt.colorbar()

plt.scatter(start_x, start_y, color='green', label='Début', zorder=5)
plt.scatter(curr_x, curr_y, color='red', label='Fin', zorder=5)

plt.axis('equal')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
