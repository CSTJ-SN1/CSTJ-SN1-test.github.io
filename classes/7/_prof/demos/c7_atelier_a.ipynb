Aire d'un cercle: PI * r^2
Aire d'un carré : 4r^2

Donc PI/4
Multiplie par 4 = ~PI

```python
import math
import random

radius = 10

ins_x = []
ins_y = []
outs_x = []
outs_y = []
total = 0

for i in range(10000000):
    x = random.uniform(-radius, radius)
    y = random.uniform(-radius, radius)
    dst = math.sqrt(x**2 + y**2)

    if dst < radius:
        ins_x.append(x)
        ins_y.append(y)
    else:
        outs_x.append(x)
        outs_y.append(y)

    total += 1
        
```

```python
import matplotlib.pyplot as plt

fig = plt.figure()
ax = fig.add_subplot()

ax.plot(ins_x, ins_y, ",")
ax.plot(outs_x, outs_y, ",")

ax.set_title("Estimation de PI")

ax.set_aspect('equal')

plt.show()
```

```python
print(len(ins_x) / total * 4)
```
