```python
import matplotlib.pyplot as plt
import random

coins_x = [0, 1, 0.5]
coins_y = [0, 0, 1]

pos_x = 0
pos_y = 0

xs = [pos_x]
ys = [pos_y]

for i in range(100000):
    coin_hasard = random.randint(0, 2)
    
    cible_x = coins_x[coin_hasard]
    cible_y = coins_y[coin_hasard]

    # Ancien ---- (nouveau) ---- Coin
    pos_x = (pos_x + cible_x) / 2
    pos_y = (pos_y + cible_y) / 2
    
    xs.append(pos_x)
    ys.append(pos_y)


# =======================

fig = plt.figure(figsize=(15, 10))
ax = fig.add_subplot()

plt.plot(xs, ys, ",")

plt.axis('off')
ax.set_title("Triagne de Sierpinski")

plt.show()
```
