### Résultat d'un dé

```python
import matplotlib.pyplot as plt
import random

resultats = [0, 0, 0, 0, 0, 0]
libelles = ["1", "2", "3", "4", "5", "6"]

for i in range(10000):
    rnd = random.randint(1, 6)
    resultats[rnd - 1] += 1


fig = plt.figure()
ax = fig.add_subplot()

ax.bar(libelles, resultats)

plt.show()
```

### Générations d'insecte

- Ces insectes vivent un total de 4 ans exactement.
- Ils deviennent adultes à partir de 1 an.
- Chaque année, les adultes produisent une nouvelle génération équivalent à 125% de la population d'adultes. (arrondi à l'inférieur)
- Après leur 4e année, ils meurent.

```python
import math

generations = [3, 0, 0, 0]

annees = 25

for i in range(annees):
    adultes = generations[1] + generations[2] + generations[3]
    nouveaux_enfants = math.floor(adultes * 1.25)

    generations[3] = generations[2]
    generations[2] = generations[1]
    generations[1] = generations[0]
    generations[0] = nouveaux_enfants
    
    print(generations)
```
