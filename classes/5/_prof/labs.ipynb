```python
import matplotlib.pyplot as plt
import math

gravity = -9.81

throw_force = 5
throw_angle = math.radians(45)

x = 0 # start position
y = 2 # initial height
vx = math.cos(throw_angle) * throw_force # initial force
vy = math.sin(throw_angle) * throw_force # horizontal throw

positions_x = []
positions_y = []

for i in range(0, 100):
    if y > 0:
        positions_x.append(x)
        positions_y.append(y)
        
    x += vx * 0.1
    y += vy * 0.1

    vy += gravity * 0.1

fig = plt.figure()
ax = fig.add_subplot()

ax.plot(positions_x, positions_y, "o-")
plt.show()
    
```

```python
import matplotlib.pyplot as plt
import math

fig = plt.figure()
ax = fig.add_subplot(projection="3d")

voxels = []

for x in range(14):
    for y in range(14):
        for z in range(14):
            # Distance
            calcul = math.sqrt((x - 7)**2 + (y - 7)**2 + (z - 7)**2)
            
            # Absolue de la distance est pas trop loin du centre?
            if abs(calcul) < 7:
                # On ajoute le voxel
                voxels.append([x,y,z])

for voxel in voxels:
    ax.bar3d(voxel[0], voxel[1], voxel[2], 1, 1, 1)

plt.show()
```
