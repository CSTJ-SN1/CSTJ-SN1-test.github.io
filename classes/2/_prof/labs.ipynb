```python
import math

a = float(input("a"))
b = float(input("b"))
c = float(input("c"))

commun = math.sqrt(b**2 - 4*a*c)
version_plus = (-b + commun) / (2*a)
version_moins = (-b - commun) / (2*a)

print("+: " + str(version_plus))
print("-: " + str(version_moins))
```
