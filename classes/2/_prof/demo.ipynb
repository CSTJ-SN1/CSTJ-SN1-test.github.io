```python
etapes_completees = 250
nombre_etapes = 500
pourcentage = etapes_completees / nombre_etapes

largeur_barre = 30
nb_plein = int(largeur_barre * pourcentage)
nb_vide = largeur_barre - nb_plein

barre = ("#" * nb_plein) + ("-" * nb_vide)

texte = "Chargement : [" + barre + "]"

print(texte)
```
