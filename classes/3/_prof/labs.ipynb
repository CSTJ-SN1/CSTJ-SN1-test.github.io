```python
n = 10

x1 = random.uniform(-n, n)
y1 = random.uniform(-n, n)
z1 = random.uniform(-n, n)
x2 = random.uniform(-n, n)
y2 = random.uniform(-n, n)
z2 = random.uniform(-n, n)

distance = math.sqrt(((x2 - x1)**2) + ((y2 - y1)**2) + ((z2 - z1)**2))

if distance > 10:
    print("Loin")
else:
    print("Proche")
    
print("Distance: " + str(distance))
```
