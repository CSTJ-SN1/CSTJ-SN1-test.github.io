```python
dessin.nouveau(500, 500, False)

dessin.couleur = "black"

for i in range(100):
    x = random.randint(0, 500)
    y = random.randint(0, 500)
    radius = random.randint(0, 50)
    dessin.cercle(x, y, radius)

dessin.afficher()
```

```python
dessin.nouveau(255, 50, False)

dessin.couleur = "black"

for x in range(255):
    dessin.couleur = (x / 255.0, x / 255.0, x / 255.0)
    dessin.rectangle(x, 0, 1, 50)

dessin.afficher()
```

```python
dessin.nouveau(300, 300, False)

dessin.couleur = "black"
dessin.couleur_bordure = "black"

x1 = 150
y1 = 150
angle = 0

while angle < math.pi * 2:
    x2 = x1 + math.cos(angle) * 200
    y2 = y1 + math.sin(angle) * 200
    dessin.ligne(x1, y1, x2, y2)
    angle += math.pi * 2 / 60.0

dessin.afficher()
```

```python
dessin.nouveau(600, 40, False)

couleurs = ["#CF1D9A", "#2DA733", "#276DBD", "#7A0D25"]
nb_couleurs = len(couleurs)

for i in range(20):
    x = i * 40

    forme = random.randint(0, 1)
    couleur_index = random.randint(0, nb_couleurs - 1)
    couleur = couleurs[couleur_index]

    dessin.couleur = couleur

    if forme == 0:
        dessin.cercle(x + 20, 20, 20)
    elif forme == 1:
        dessin.rectangle(x, 0, 40, 40)


dessin.afficher()
```
