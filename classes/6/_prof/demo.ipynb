```python
import math
import csv
import matplotlib.pyplot as plt

xs = []
ys = []

with open('demo_files/Photo_nord_quebecois.csv') as f:
    lignes = csv.reader(f, delimiter=';')

    next(lignes)
    
    for ligne in lignes:
        # FID;PHOTO_URL;LATITUDE;LONGITUDE;ALTITUDE;DIRECTION;DATUM;SYMBOLOGIE;DATE;NOM_PHOTO
        lat_str = ligne[2]
        lon_str = ligne[3]

        lat = float(lat_str.replace(",", "."))
        lon = float(lon_str.replace(",", "."))

        # https://en.wikipedia.org/wiki/Mercator_projection
        lon_fixed = math.cos(math.radians(58)) * lon

        xs.append(lon_fixed)
        ys.append(lat)

fig = plt.figure()
ax = fig.add_subplot()

ax.axis('off')
ax.plot(xs, ys, ",")

plt.show()
```
