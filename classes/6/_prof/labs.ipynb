```python
import csv

total = 0
qty = 0

with open("liste_a.csv") as f:
    lignes = csv.reader(f)
    
    for ligne in lignes:
        n = int(ligne[0])
        total += n
        qty += 1

    print(qty)
    print(total / qty)
```

```python
import csv

plus_petit = 999999
plus_grand = 0

with open("liste_a.csv") as f:
    lignes = csv.reader(f)
    
    for ligne in lignes:
        n = int(ligne[0])

        if n < plus_petit:
            plus_petit = n
            
        if n > plus_grand:
            plus_grand = n

    print(plus_grand)
    print(plus_petit)
    print(plus_grand - plus_petit)
```
