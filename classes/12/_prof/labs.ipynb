```python
nb = int(input())
b = format(nb, "b")

ones = 0
zeros = 0

for c in b:
    if c == "0":
        zeros += 1
    else:
        ones += 1

print(f"Le binaire pour {nb} est {b}. Il contient {zeros} zéros et {ones} uns")
```

```python
def fn1(s):
    return s[0]

assert fn1("Allo") == "A"
assert fn1("Mais que fait") == "M"
assert fn1("Cette fonction") == "C"
assert fn1("?!!!!") == "?"
print("Succès")
```

```python
def fn2(a, b):
    if a > b:
        return a
    return b

assert fn2(10, 5) == 10
assert fn2(20, 100) == 100
assert fn2(-2, 2) == 2
assert fn2(5, 1) == 5
print("Succès")
```

```python
def fn3(nb, qte_zeros):
    s = format(nb, "b")
    for i in range(qte_zeros):
        s += "0"
    nb2 = int("0b" + s, 2)
    return nb2

assert fn3(42, 1) == 84
assert fn3(42, 2) == 168
assert fn3(0, 215125) == 0
assert fn3(1, 16) == 65536
assert fn3(115521, 2) == 462084
print("Succès")
```
